```python
import math


def max_forward_accel(pc):
    """
    Maximum forward acceleration at zero airspeed.

    a_forward_max = T_para_max / mass
    """
    return pc['T_para_max'] / pc['mass']


def max_lateral_accel(pc):
    """
    Maximum lateral / centripetal acceleration.

    a_lateral_max = T_perp_max / mass
    """
    return pc['T_perp_max'] / pc['mass']


def max_forward_airspeed(pc):
    """
    Terminal forward airspeed in still air.

    v_forward_inf = T_para_max / drag
    """
    return pc['T_para_max'] / pc['drag']


def max_angular_accel(pc):
    """
    Maximum angular acceleration from rest.

    alpha_max = tau_max / inertia
    """
    return pc['tau_max'] / pc['inertia']


def max_angular_velocity(pc):
    """
    Terminal angular velocity under maximum yaw torque.

    omega_inf = tau_max / k_rot
    """
    return pc['tau_max'] / pc['k_rot']


def max_coordinated_turn_rate(pc, airspeed):
    """
    Maximum yaw rate that lateral thrust can support at a given airspeed,
    assuming a coordinated turn with negligible sideslip.

        T_perp = m * v * omega

        omega_max = T_perp / (m * v)

    Returns rad/s.
    """
    if airspeed <= 0:
        return math.inf

    return pc['T_perp_max'] / (pc['mass'] * airspeed)


def translational_time_constant(pc):
    """
    tau_lin = mass / drag
    """
    return pc['mass'] / pc['drag']


def rotational_time_constant(pc):
    """
    tau_rot = inertia / k_rot
    """
    return pc['inertia'] / pc['k_rot']
```
